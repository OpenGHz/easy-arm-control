```python
import rospy
from airbot_play_control.control import RoboticArmAgent
# ROS节点初始化
NODE_NAME = 'airbot_play_gazebo'
rospy.init_node(NODE_NAME)
```

```python
# 初始化机械臂本体
arm = RoboticArmAgent(node_name=NODE_NAME, other_config = ("","airbot_play_arm"))
```

```python
arm.get_current_pose()
```

```python
# 控制机械臂末端到达指定的工作空间位置
position = [0.4,0,0.3]  # xyz位置值
rotation = [-3.14,0,0]  # rpy角度值或xyzw四元数
arm.set_and_go_to_pose_target(position,rotation,target_base='0',sleep_time=2)
```

```python
# 机械臂本体控制（详见：example_control.ipynb）
arm.go_to_named_or_joint_target('Home')
```

```python
# 初始化夹爪(夹爪只能使用关节空间控制，因为是闭链结构，无法使用逆解)
gripper = RoboticArmAgent(node_name=NODE_NAME, other_config = ("","airbot_play_gripper"))
```

```python
# 夹爪闭合
SCALE_CLOSE = 0.8
CLOSE = (-SCALE_CLOSE,SCALE_CLOSE,SCALE_CLOSE,-SCALE_CLOSE)
gripper.go_to_named_or_joint_target(CLOSE)
```

```python
# 夹爪张开
SCALE_OPEN = 0
OPEN = (SCALE_OPEN,SCALE_OPEN,SCALE_OPEN,SCALE_OPEN)
gripper.go_to_named_or_joint_target(OPEN)
```
