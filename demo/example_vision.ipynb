# Example Vision-视觉调参

### 初始化

```python
from airbot_play_control.control import RoboticArmAgent, ChooseGripper
import rospy

# ROS节点初始化
NODE_NAME = "airbot_play_vision_hsv"
rospy.init_node(NODE_NAME)
other_config = None
sim_player_nums = rospy.get_param("/airbot_play/sim_player_nums", default=0)
sim_type = ["real", "gazebo"][sim_player_nums]
print("Use", sim_type)
gripper = ["gripper", "inspire"][0]
if sim_type == "gazebo":
    other_config = ("", "airbot_play_arm")
gripper_controller = ChooseGripper(sim_type, gripper)
arm = RoboticArmAgent(
    node_name=NODE_NAME, gripper=(4, gripper_controller), other_config=other_config
)
if sim_type == "gazebo":
    arm.set_and_go_to_pose_target([0.2739,0,-0.27],[0,1.570796,0],sleep_time=2)
```

### HSV整定

```python
from std_vision.video import StdVideo

# 执行如下代码可以通过滑动条调整视觉参数
# 按ESC键退出
if sim_type != "real":
    HSV_L, HSV_H = StdVideo.color_thresh_determine("/camera/color/image_raw")
else:
    HSV_L, HSV_H = StdVideo.color_thresh_determine(2)
```

#### 退出ROS节点

```python
# 关闭视觉节点
rospy.signal_shutdown('end')
```
