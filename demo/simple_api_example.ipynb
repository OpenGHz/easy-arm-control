#### 预配置

```python
import rospy

rospy.set_param("/airbot_play/sim_player_nums", 1)  # 1表示使用仿真，0表示使用真机
```

#### 整体初始化

```python
import rospy
from easy_arm_control.control import RoboticArmAgent, GripperSimpleControl

NODE_NAME = "airbot_play_uni_control"
rospy.init_node(NODE_NAME)
arm = RoboticArmAgent(node_name=NODE_NAME)
gripper = GripperSimpleControl()
```

#### 夹爪控制

```python
gripper.control(1)  # 1/True闭合
gripper.get_status()
```

```python
gripper.control(0)  # 0/False张开
gripper.get_status()
```

#### 机械臂本体控制

```python
# 查询目标位置是否可达
print(arm.check_pose((1, 0, -0.094, 0, 0.698, 0, 0.716)))
print(arm.check_pose((0.4, 0, 0.3, -3.14, 0, 0)))
```

```python
# 获取关节状态
current_joint_states = arm.get_current_joint_values()
# 获取末端位姿（相对base_link）
current_pose = arm.get_current_pose()
print(current_joint_states, "\n", current_pose)
```

```python
# 移动到单一目标位置
arm.set_and_go_to_pose_target([0.3472, 0.0796, -0.231, 0, 0.7, 0, 0.714])
```

```python
# 给定一系列路点移动
from copy import deepcopy

base_pose = list(arm.get_current_pose())
pose1 = deepcopy(base_pose)
pose1[0] -= 0.02
pose2 = deepcopy(pose1)
pose2[1] -= 0.02
arm.set_and_go_to_way_points([pose1, pose2], start_base=0)
```
